```python
# set runtime to T4 GPU
from transformers import pipeline

# Load the fine-tuned BERT-Emotion model
sentiment_analysis = pipeline("text-classification", model="boltuix/bert-emotion")

# Analyze emotion
result = sentiment_analysis("i love you")
print(result)
```

```python
sentiment_analysis(["this product does not work well", "i like that product"])
```

```python
import pandas as pd
```

```python
df = pd.read_csv("https://github.com/christianw/applied-ux/raw/main/week03/amazon_reviews.csv.xz", low_memory=False)
```

```python
pos_mask = df["name"] == "Fire Tablet, 7 Display, Wi-Fi, 8 GB - Includes Special Offers, Magenta"
```

```python
pos = df[pos_mask].copy()
```

```python
len(pos)
```

```python
pos.dtypes
```

```python
sentiments = sentiment_analysis(list(pos["reviews.title"].map(str)))
```

```python
pos["sentiment"] = [s["label"] for s in sentiments]
```
